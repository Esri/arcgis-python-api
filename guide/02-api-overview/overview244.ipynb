# Summary

See below for brief introductions to some new features introduced with the ArcGIS Python API 2.4.4 release. Also, checkout the [What's new in ArcGIS API for Python 2.4.4](https://www.esri.com/arcgis-blog/products/api-python/announcements/whats-new-in-arcgis-api-for-python-v2-4-4) blog for additional areas updated in the API. For a comprehensive list of changes since the 2.4.3 release, visit the [Release Notes for 2.4.4](../release-notes-244) document.

## Release Highlights

### Dependency Management changes
A new high-performance [Rust](https://rust-lang.org/) geometry library is shipped with the Python API 2.4.4, laying the foundation for faster performance and a wider range of automation capabilities for content management, analysis, and administration workflows. Just import from the new `arcgis_geometry` package rather than the previous `arcgis.geoemtry` module, and take advantage of these capabilities. Then, just run your code and reap the benefits.

### Additional functions in the `arcgis.ai` module to work with beta ArcGIS Online services
> **Note:** These new functions in the *arcgis.ai* module currently depend on ArcGIS Online services that are still in beta and are expected to transition to general availability in a future release. These services may also incur ArcGIS Online credit usage for transactions made through the arcgis.ai module once they graduate from beta to general release in the future.

In addition to the functions that have transitioned to *General Availablity*, [analyze_text()](/python/latest/api-reference/arcgis.ai.html#analyze-text), [translate()](/python/latest/api-reference/arcgis.ai.html#translate), and [analyze_image()](/python/latest/api-reference/arcgis.ai.html#analyze-image), the 2.4.4 release introduces four new **beta** functions to explore AI capabilities:
* [`image_exract_text()`](/python/latest/api-reference/arcgis.ai.html#image-extract-text) - to pull out text from image files
* [`generate_speech()`](/python/latest/api-reference/arcgis.ai.html#generate-speech) - create speech from text data
* [`transcribe_audio()`](/python/latest/api-reference/arcgis.ai.html#transcribe-audio) - write out what's recoreded in an audio file 
* [`analyze_file()`](/python/latest/api-reference/arcgis.ai.html#analyze-file) - examine the contents of files and documents
These capabilities will not just be available for use with ArcGIS Online organizations, but will also be accessible with the ArcGIS Enterprise 12.2 release. Esri operates these services so you can use the Python API to take advantage of their analysis and processing abilities.

### Improvements to GIS Administration and Content Management
Performance and efficiency improvements abound at the 2.4.4 reelase, including numerous updates to internal processing functions to avoid unnecessary and time-consuming calls to external services, as well as eliminating operations that tried to resolve private urls when security mechanisms did not require them. The new [UserManager.all()](/python/latest/api-reference/arcgis.gis.toc.html#arcgis.gis.UserManager.all) method provides the capability to return all *users* in an organization even when that number exceeds the [10000 limit](/rest/users-groups-and-items/considerations-and-limitations/#search-limitations) for pagination.  Major updates to the [OfflineContentManager](/python/latest/api-reference/arcgis.gis.toc.html#offlinecontentmanager) introduce the capability to export and import content asynchronously to avoid idle time while processing large amounts of content or extremely complicated items. Since it integrates with the processing capabilities of the [arcgis.apps.itemgraph](/python/latest/api-reference/arcgis.apps.itemgraph.html) module, services are validated and included within internal processing dependency graphs, providing more accurate representations of how items, applications, and services are connected. In addition, now *KnowledgeGraph* items are explicitly supported with this class to enable cloning of these items. Various *LogManager.query()* methods were updated in numerous *arcgis.gis* submodules, including [arcgis.gis.mission](/python/latest/api-reference/arcgis.gis.mission.html) and [arcgis.gis.nb](/python/latest/api-reference/arcgis.gis.nb.html), to allow for more options when specifying *start_time* and *end_time* values. New classes, methods, and models were added to the [arcgis.graph](/python/latest/api-reference/arcgis.graph.html) module to enable synchronization workflows with *KnowledgeGraphs*.

### Enhancements in Mapping Capabilities and Visualization
An reorganizaiton of the *SmartMapping* functionality presents new documentation subheadings within the [arcgis.map](/python/latest/api-reference/arcgis.map.toc.html) module. The [Smart Mapping](/python/latest/api-reference/arcgis.map.toc.html#smart-mapping) subheading groups the [`SmartMappingManager`](/python/latest/api-reference/arcgis.map.toc.html#smartmappingmanager) and various data models (objects which function like classes) aimed at simplifying inputs for visualizing data on maps while also provided validation before runtime to avoid errors. The [Smart Mapping Enumerations](/python/latest/api-reference/arcgis.map.toc.html#smart-mapping-enumerations) provides member objects to assist in the creation of new *RendererParam* objects for input to the [`SmartMappingManager.create_renderer()`](/python/latest/api-reference/arcgis.map.toc.html#arcgis.map.smart_mapping.SmartMappingManager.create_renderer) method so as to inject data validation before sending requests to the server for quicker error catching. The goal of introducing these new classes is to simplify the creation of the numerous rendering options the Python API *map widget* provides through its integration with the latest release of the ArcGIS Maps SDK for Javascript.

### Updates to Core GIS Functionality
Enhanced documentation, particularly with more code examples for initializing class objects and analysis functions, increases the accuracy of code results produced from *ai agents* being asked questions about the Python API. See the [FeatureLayer](/python/latest/api-reference/arcgis.features.toc.html#featurelayer), [FeatureLayerCollection](/python/latest/api-reference/arcgis.features.toc.html#featurelayercollection), [ParcelFabricManager](/python/latest/api-reference/arcgis.features.managers.html#parcelfabricmanager), and various [arcgis.features.analysis](/python/latest/api-reference/arcgis.features.analysis.html) API reference pages for examples of the type of updates being made to increase accuracy of AI agent code. The [ContentManager.create_service()](/python/latest/api-reference/arcgis.gis.toc.html#arcgis.gis.ContentManager.create_service) and [Item.update()](/python/latest/api-reference/arcgis.gis.toc.html#arcgis.gis.Item.update) methods have been enhanced to allow sharing with groups when the *item_properties* argument is configured appropriately. Updates to the [`SharingManager`](/python/latest/api-reference/arcgis.gis.toc.html#sharingmanager) have improved performance when accessing and setting properties. In addition, modifications to the [`Folder`](/python/latest/api-reference/arcgis.gis.toc.html#folder) class's [`add()`](/python/latest/api-reference/arcgis.gis.toc.html#arcgis.gis._impl._content_manager.Folder.add) and [`list()`](/python/latest/api-reference/arcgis.gis.toc.html#arcgis.gis._impl._content_manager.Folder.list) methods create and retrieve content more reliably and quickly.

## Additional Details

### Documentation details for changes in the [arcgis.graph](/python/latest/api-reference/arcgis.graph.html) module
The new [Replica-Synchronization Types](/python/latest/api-reference/arcgis.graph.html#replica-synchronization-types) API Reference section of the module fully documents new classes for working with inputs and responses of the new [`KnowledgeGraph.create_replica()`](/python/latest/api-reference/arcgis.graph.html#arcgis.graph.KnowledgeGraph.create_replica), [`KnowledgeGraph.sychronize_replica()`](/python/latest/api-reference/arcgis.graph.html#arcgis.graph.KnowledgeGraph.synchronize_replica), [`KnowledgeGraph.get_replica_definition()`](/python/latest/api-reference/arcgis.graph.html#arcgis.graph.KnowledgeGraph.get_replica_definition), and [`KnowledgeGraph.unregister_replica()`](/python/latest/api-reference/arcgis.graph.html#arcgis.graph.KnowledgeGraph.unregister_replica) methods.


### Transition of the arcgis.raster.realitymapping module into General Availability with ArcGIS Enterprise 12.2
The [arcgis.raster.realitymapping](/python/latest/api-reference/arcgis.raster.realitymapping.html) module will function reliably when [ArcGIS Reality Server](https://doc.esri.com/en/arcgis-reality-server/latest/introduction/what-is-reality-server.html) transitions to General Availability with the release of ArcGIS Enterprise 12.2. Classes, properties, and methods will not change between releases such to cause any breaking change in code.

### New methods and extensive updates in the [arcgis.layers](/python/latest/api-reference/arcgis.layers.toc.html) module
The 2.4.4 release focused on providing fully functional caching workflows with the introduction of new methods for cache and tile management to the [`EnterpriseMapImageLayerManager`](/python/latest/api-reference/arcgis.layers.toc.html#arcgis.layers.EnterpriseMapImageLayerManager) class.  Documentation updates elaborate on initialization of *Map Service* layer objects, and add narrative descriptions to better detail concepts and usage for the classes within the module.
