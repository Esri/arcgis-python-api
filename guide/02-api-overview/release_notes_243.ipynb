# What's new in version 2.4.3

## New
* Updates documentation regarding [ArcGIS Online terms of use](https://doc.arcgis.com/en/arcgis-online/reference/terms-of-use.htm)
* **New modules:**
  * [`arcgis.ai`](/python/latest/api-reference/arcgis.ai.html)
    * Provides functionality for image and text analysis and translation 
  * [`arcgis.apps.dashboards`](/python/latest/api-reference/arcgis.apps.dashboards.html)
    * Focused on managing content dependencies within Dashboard items for more fluid cloning capability
* **Deprecated modules:**
  * [`arcgis.apps.dashboard`](/python/latest/api-reference/arcgis.apps.dashboard.html)
    * Removal of the module, which has seen no active development since the 2.1.0 release.
