# Introduction to Feature Analysis

Feature analysis is the process of performing server-side spatial analysis operations on feature data such as points, lines or polygons. These operations can be used to find features, merge or overlay features, calculate feature statistics, or find patterns and relationships in feature data. The operation you use depends on the type of problem you are trying to solve.

Feature analysis can help solve different types of problems. For example, you can use it to:

* Understand where features are or where multiple features exist.
* Measure sizes, shapes, and distributions of features.
* Analyze relationships and interactions between features.
* Detect and quantify patterns and relationships between features.
* Make predictions based on existing or theoretical patterns and relationships.


### Performing Feature Analysis using the ArcGIS API for Python


To perform an analysis, you can use a hosted feature layer as input for the analysis tool. The feature analysis service can access this hosted data and execute operations and store results either as a hosted feature layer or features in-memory. 

Within the context of the ArcGIS API for Python, these analysis tools can be found under the [`features.analysis`](/python/latest/api-reference/arcgis.features.analysis.html) submodule as well as other [`submodules of the _features_ module](/python/latest/api-reference/arcgis.features.toc.html#submodules). The following are valid __input formats__ for analysis tools, unless mentioned otherwise for a specific tool in the [API reference](/python/latest/api-reference/) for that tool.

1. [`FeatureCollection`](/python/latest/api-reference/arcgis.features.toc.html#arcgis.features.FeatureCollection) - an in-memory collection of [`Feature`](python/latest/api-reference/arcgis.features.toc.html#arcgis.features.Feature) objects with rendering information.
2. [`FeatureLayer`](/python/latest/api-reference/arcgis.features.toc.html#arcgis.features.FeatureLayer) - represents an individual layer within a [feature service](https://doc.esri.com/en/arcgis-enterprise/latest/administer/what-is-a-feature-service-.html).
3. [`FeatureLayerCollection`](/python/latest/api-reference/arcgis.features.toc.html#arcgis.features.FeatureLayerCollection) - a collection of feature layers and tables, the Python API representation of a feature service.
4. [`Item`](/python/latest/api-reference/arcgis.gis.toc.html#arcgis.gis.Item) of type `FeatureCollection` or `FeatureLayerCollection` where only the first layer is used.


The feature analysis tools usually return results of the following __output formats__, unless mentioned otherwise for a specific tool. 

1. [`FeatureLayer`](/python/latest/api-reference/arcgis.features.toc.html#featurelayer) - hosted feature layer in the ArcGIS Organization you have authenticated into, if `output_name` is provided.
2. [`FeatureCollection`](python/latest/api-reference/arcgis.features.toc.html#featurecollection) - result can be accessed in-memory.

### Example of using the tools and visualizing results

Let's look at an example that finds specific data of interest, finding hot spots within that data, and displays the results of the analysis on a map. 

```python
from arcgis.gis import GIS
from arcgis.features.analyze_patterns import find_hot_spots
```

```python
gis = GIS(profile='your_online_profile')
```

We access data for Crime incidents in Washington, DC for the first half of 2026 through the itemid for the hosted *feature layer item*, represented in the Python API as a [Feature Layer Collection](/python/latest/api-reference/arcgis.features.toc.html#featurelayercollection).

```python
input_data = gis.content.get("5173a6923bc140a58afcab50356c3439")
input_data
```

```python
input_data_layer = input_data.layers[0]
type(input_data_layer)
```

We verify that a [`FeatureLayer`](/python/latest/api-reference/arcgis.features.toc.html#featurelayer] object was extracted from the item. Then we run this tool with a desired  name for the resultant layer as the *output_name* argument. The output item will be created in our GIS Organization.

```python
output_name = "crime_incidents_dc_2026_hotspots"
hot_spots_result = find_hot_spots(analysis_layer = input_data_layer, output_name = output_name)
hot_spots_result
```

We will now visualize these results in a map. 

```python
# Create a map centered on Washington DC
map1 = gis.map("Washington, DC")

# Add results to the map
map1.content.add(hot_spots_result)
# Set the basemap 
map1.basemap.basemap = 'arcgis-dark-gray'
# Visualize
map1
```
