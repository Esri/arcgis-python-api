## Operations
With the geometry objects above, run the `buffer` and `buffer_collection` operations

```python
multipoint_buffer = multipoint.buffer(distance=125.9)
print("Area of multipoint buffer:", multipoint_buffer.area())
render_geom(multipoint_buffer, "polygon")
```
