### Extract text from image

![image](https://geosaurus.maps.arcgis.com/sharing/rest/content/items/36c34ac9e2c5402ab94762f97f3349af/data)

```python
virginia_map_url = "https://geosaurus.maps.arcgis.com/sharing/rest/content/items/36c34ac9e2c5402ab94762f97f3349af/data"
```

```python
prompt_list = [
            "Extract the title from the image.",
            "Find the state names in this image.",
        ]
```

```python
result_list = image_extract_text(image = virginia_map_url, prompt = prompt_list)
```

```python
result_list.results[0].value
```

```python
result_list.results[1].value
```
