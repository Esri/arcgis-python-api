# Create a Map Showcase instance from existing Item

```python
item = gis.content.get("90d766014e7e48c883544b23e2cffd81")
map_showcase = MapShowcase(item)
map_showcase
```

## `MapShowcase` methods and properties
A Map Showcase contains Map Showcase Items. These can be various item types. Each item within a Map Showcase can have it's own properties so we have created a separate `MapShowcaseItem` page to represent them.

A `MapShowcaseItem` can have a thumbnail, title, renderer, etc. that are defined only within the Map Showcase.
