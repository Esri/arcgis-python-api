# Spatially join points and polygon - Local Files


Spatially join or aggregate the number of points over each polygon in a layer and view the results in a map within the notebook. 

> NOTE: This notebook demonstrates the pattern where points and polygons are being accessed through __local files (.csv or .shp) in your notebook workspace__.
>
> If you wish to implement this workflow with data accessed through __hosted service urls__, please refer to the Template titled "Spatially Join points and polygon - Hosted Services".


## Import necessary libraries and connect to GIS


```python
from arcgis.gis import GIS
from arcgis.features import FeatureLayer
import pandas as pd
```

```python
gis = GIS("home")
```

## 1. Define inputs


```python
#pattern for accessing data through both .csv or .shp files
path_to_points_data = '<PATH-TO-YOUR-POINTS-CSV>' 
path_to_polygon_data = '<PATH-TO-YOUR-POLYGON-SHP>'

#coordinate_system to access data in
coordinate_system_id = <WELL_KNOWN_ID> #integer e.g. 4326

#column names for the .csv file
points_data_x_column_name = '<COLUMN-NAME-FOR-LONGITUDE-COLUMN>'
points_data_y_column_name = '<COLUMN-NAME-FOR-LATITUDE-COLUMN>'
```

## 2. Read in the .CSV points data


```python
df1 = pd.read_csv(path_to_points_data)
sedf_points = pd.DataFrame.spatial.from_xy(df1, x_column = points_data_x_column_name, y_column = points_data_y_column_name, sr = coordinate_system_id)
sedf_points.head()
```

## 3. Read in the .SHP polygon data


```python
sedf_polygon = pd.DataFrame.spatial.from_featureclass(location=path_to_polygon_data)
sedf_polygon.head()
```

## 4. Checking the coordinate systems for the two Spatially Enabled DataFrames


This step ensures they are compatible for spatial operations.

```python
sr_points = sedf_points.spatial.sr
sr_points
```

```python
sr_polygon = sedf_polygon.spatial.sr
sr_polygon
```

### 4.1 Reprojecting one of the two datasets (if needed) to spatially join them in the same coordinate system


Skip this step and proceed to the next if both the datasets have identical values for `sr`.

```python
sedf_points.spatial.project(spatial_reference=sr_polygon)
```

## 5. Spatially joining the two Spatially Enabled DataFrames


```python
joined_sedf = sedf_polygon.spatial.join(sedf_points, how='left')
joined_sedf.head()
```

## 6. Plotting on a map


```python
map2 = gis.map()
joined_sedf.spatial.plot(map2)
map2.zoom_to_layer(joined_sedf)
map2
```
