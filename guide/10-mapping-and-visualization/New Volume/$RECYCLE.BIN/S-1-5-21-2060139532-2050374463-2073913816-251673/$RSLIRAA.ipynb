# Summarize data code snippets


#### Run this cell to connect to your GIS and get started:

```python
from arcgis.gis import GIS
gis = GIS("home")
```

#### Now you are ready to start!

### Aggregate points

```python
from arcgis.features.summarize_data import aggregate_points

# Set inputs
point_itemid = "<ITEM-ID-FOR-POINT-LAYER-ITEM>"
polygon_itemid = "<ITEM-ID-FOR-POLYGON-LAYER-ITEM>"
output_name = "<TITLE-FOR-ANALYSIS-OUTPUT>"

# Access data
point_layer = gis.content.get(point_itemid).layers[0]
polygon_layer = gis.content.get(polygon_itemid).layers[0]

# Perform analysis
result = aggregate_points(point_layer = point_layer, polygon_layer = polygon_layer, output_name = output_name)

# Visualize results
map_aggregate = gis.map()
map_aggregate.content.add(result)
map_aggregate.zoom_to_layer(result)
map_aggregate
```

### Join Features

```python
from arcgis.features.summarize_data import join_features

# Set inputs
target_layer_itemid = "<ITEM-ID-FOR-TARGET-LAYER-ITEM>"
join_layer_itemid = "<ITEM-ID-FOR-JOIN-LAYER-ITEM>"
output_name = "<TITLE-FOR-ANALYSIS-OUTPUT>"
# set attribute_relationship = None, and provide values for spatial_relationship, spatial_relationship_distance, spatial_relationship_units instead if you prefer to perform spatial join
attribute_relationship = [{"targetField": "<TARGET-FIELD-NAME>", "operator": "<OPERATOR>", "joinField": "<JOIN-FIELD-NAME>"}] 

# Access data
target_layer = gis.content.get(target_layer_itemid).layers[0]
join_layer = gis.content.get(join_layer_itemid).layers[0]

# Perform analysis - add parameters for spatial join if they have been defined above
result = join_features(target_layer = target_layer, join_layer = join_layer, output_name = output_name, attribute_relationship = attribute_relationship)

# Visualize results
map_join = gis.map()
map_join.content.add(result)
map_join.zoom_to_layer(result)
map_join
```

### Summarize center and dispersion

```python
from arcgis.features.summarize_data import summarize_center_and_dispersion

# Set inputs
analysis_layer_itemid = "<ITEM-ID-FOR-ANALYSIS-LAYER-ITEM>"
# choose one or more summarize_type values from these accepted values -  [“CentralFeature”, “MeanCenter”, “MedianCenter”, “Ellipse”] - and provide it as a list
summarize_type_value = ["Ellipse"]
output_name = "<TITLE-FOR-ANALYSIS-OUTPUT>"

# Access data
analysis_layer = gis.content.get(analysis_layer_itemid).layers[0]

# Perform analysis
result = summarize_center_and_dispersion(analysis_layer = analysis_layer, summarize_type = summarize_type_value, output_name = output_name)

# Visualize results
map_summarize = gis.map()
map_summarize.content.add(result)
map_summarize.zoom_to_layer(result)
map_summarize
```

### Summarize Nearby

```python
from arcgis.features.summarize_data import summarize_nearby

# Set inputs
sum_nearby_layer_itemid = "<ITEM-ID-FOR-SUM-NEARBY-LAYER-ITEM>"
summary_layer_itemid = "<ITEM-ID-FOR-SUMMARY-LAYER-ITEM>"
# if you wish to update the 'near_type' (mesurement parameter) please do that below, else you may leave the default value as is. Accepted values for near_type are listed below.
# near_type: 'StraightLine', 'Driving Distance', 'Driving Time', 'Rural Driving Distance', 'Rural Driving Time', 'Trucking Distance', 'Trucking Time', 'Walking Distance', 'Walking Time'
near_type = "StraightLine"
# provide search distances as a list of value(s) below by updating the preset default value
distances = [100]
# If you wish to update the 'units' for distances set above, please to do that below. Acceptable values for units: Meters, Kilometers, Feet, Yards, Miles
units = "Meters"
output_name = "<TITLE-FOR-ANALYSIS-OUTPUT>"

# Access data
sum_nearby_layer = gis.content.get(sum_nearby_layer_itemid).layers[0]
summary_layer = gis.content.get(summary_layer_itemid).layers[0]

# Perform analysis
result = summarize_nearby(sum_nearby_layer = sum_nearby_layer, summary_layer = summary_layer, near_type = near_type, distances = distances, output_name = output_name)

# Visualize results
map_nearby = gis.map()
map_nearby.content.add(result)
map_nearby.zoom_to_layer(result)
map_nearby
```

### Summarize Within

### Zonal statistics

```python
from arcgis.raster.analytics import summarize_raster_within

# Set inputs
input_zone_layer_itemid = "<ITEM-ID-FOR-INPUT-ZONE-LAYER>" # raster or feature data
input_raster_layer_to_summarize_itemid = "<ITEM-ID-FOR-INPUT-RASTER-LAYER-TO-SUMMARIZE>"
output_name = "<TITLE-FOR-ANALYSIS-OUTPUT>"

# Access data
input_zone_layer = gis.content.get(input_zone_layer_itemid).layers[0]
input_raster_layer_to_summarize = gis.content.get(input_raster_layer_to_summarize_itemid).layers[0]

# Perform analysis
result = summarize_raster_within(input_zone_layer = input_zone_layer, input_raster_layer_to_summarize = input_raster_layer_to_summarize, output_name = output_name)

# Visualize results
map_raster_within = gis.map()
map_raster_within.content.add(result)
map_raster_within.zoom_to_layer(result)
map_raster_within
```
