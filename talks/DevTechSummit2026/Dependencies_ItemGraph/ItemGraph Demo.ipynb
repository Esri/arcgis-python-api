## The ItemGraph
### What is it, how does it work, and how do I use it?

The module currently consists of 2 classes and 2 functions:
- ItemGraph
- ItemNode
- create_dependency_graph()
- load_from_file()

We'll take a brief look at the [doc](https://developers.arcgis.com/python/latest/api-reference/arcgis.apps.itemgraph.html) for these.

#### Customizable
The ItemGraph is an extension of NetworkX's [DiGraph class](https://networkx.org/documentation/stable/reference/classes/digraph.html), giving users all of its latent functionality. Additionally, the ItemGraph class includes functions for adding/deleting items and relationships to/from a graph. We'll take a look at some applications of both of these perks a bit later.

#### Troubleshooting

One of the most helpful parts of this functionality is its ability to identify broken data sources (a dependency that doesn't exist). It does this a few different ways, such as searching in specific spots of application structures and doing a cursory check over all parts of structures for formats matching Item Id's. In certain cases, it can include service URL's as well.

The ItemGraph will attempt to retrieve the item associated with the ID or the service associated with the URL. If this fails, then the item will be included in the graph as an "Unknown Type," and the ItemNode won't have anything stored in its .item property. Checking your graph for nodes of these type is a quick way to identify data sources that made need fixing in your applications.

```python
fl_node.type
```

```python
for node in survey_graph.all_items():
    if node.type == "Unknown Type":
        print(node.id)
```

```python
review_items = set()
for node in survey_graph.all_items():
    if node.type == "Unknown Type":
        review_items.update(node.contained_by("item"))
review_items
```
