## clone_items()

```python
from arcgis.gis import GIS
gis1 = GIS(profile="your_online_profile")
gis2 = GIS(profile="np_pypi")
```

```python
source_map = gis1.content.get("bf5a039b1947422190b5812fc34af717")
source_map
```

```python
source_map.get_dependencies()
```

```python
cl = gis2.content.clone_items([source_map], folder = "cloning_demo", search_existing_items = False)
```

```python
cl
```

```python
offline_path = gis1.content.offline.export_items([source_map])
offline_path
```

```python
gis2.content.offline.list_items(offline_path)
```

```python
gis2.content.offline.import_content(offline_path, folder = "offline_cloning_demo")
```
